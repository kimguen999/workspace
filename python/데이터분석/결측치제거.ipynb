```python
df = pd.read_excel('teet2.xlsx')
df
```

```python
# 결측치 : 데이터 세트에서 누락된 값(비어있는 데이터, NaN)
# 결측치 제거 : 비어있거나 NaN 데이터를 삭제(행 or 열 삭제)
#              비어있거나 NaN 데이터를 특정 데이터로 채워줌
# 시험문제
```

```python
# 실습 위해 df에 결측치 추가
import numpy as np

df['학급'] = np.nan
df
```

```python
# 결측치를 특정 데이터로 채우기

df1 = df.copy()

# fillna 함수는 nan 데이터에 특정 데이터를 채운 결과를 리턴해줌(원본 데이터는 변경사항 없음)
result_df = df1.fillna('임시데이터')

# 결측치가 래워진 데이터프레임
result_df

# df1은 결측치기 남아있음(NaN)
df1
```

```python
# 특정 컬럼의 결측치 제거

df1 = df.copy()
#취미 컬럼의 결측치만 특정 데이터로 채우기
df1['취미'].fillna('없음')

# df1을 출력하면 결측치가 남아있음
df

df1['취미'].fillna('없음', inplace=True)
df1
```
