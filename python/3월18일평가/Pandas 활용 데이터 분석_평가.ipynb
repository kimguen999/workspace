1. 키보드로 세 수를 입력받고,입력받은 세 수를 리스트에 저장하시오. 그 후 리스트에 저장된 수 중 가장 큰 수와 가장 작은 수의 차를 출력하시오.

```python
num1 = int(input("정수입력 : "))
num2 = int(input("정수입력 : "))
num3 = int(input("정수입력 : "))
max = 0
min = 0
if num1>num2 and num1>num3 :
  max = num1
elif num2>num3 and num2>num1 :
  max = num2
else :
  max = num3

if num1<num2 and num1<num3 :
  min = num1
elif num2<num3 and num2<num1 :
  min = num2
else :
  min = num3

char = max - min
print(f"첫번째 입력박은 정수 : {num1}")
print(f"두번째 입력박은 정수 : {num2}")
print(f"세번째 입력박은 정수 : {num3}")
print(f"입력받은 세 정수 중 가장 큰 수와 가장 작은 수의 차는 {char}입니다.")
```

2. 매개변수로 두 개의 리스트를 받아, 두 개의 리스트에 저장된 모든 데이터의 합을 리턴하는 함수를 작성하시오.  
EX> 두 매개변수 : [1,2,3] , [4,5,6] / 결과 : 21

```python
list1 = [1,2,3]
list2 = [4,5,6]
def sumList(list1, list2):
  return sum(list1) + sum(list2)
print(sumList(list1, list2))
```

3. 매개변수로 두 개의 리스트를 받아, 두 개의 리스트에 중복으로 저장된 데이터만을 갖는  새로운 리스트를 리턴하는 함수를 작성하시오.  
EX> 두 매개변수 : [1,3,5,7,9] , [1,2,3,4,5] / 결과 : [1,3,5]

```python
list1 = [1,3,5,7,9]
list2 = [1,2,3,4,5]

def jungbok(list1, list2):
  jungbokResult = []
  for e in list1:
    if e in list2:
      jungbokResult.append(e)
  return jungbokResult
print(jungbok(list1,list2))
```

6. 아래 딕셔너리 데이터에 stu_grade라는 키를 추가하고, 해당 키의  데이터로  점수(stu_score)에 따른 등급 데이터를 저장하시오.  
90 <= 점수      : A등급  
80 <= 점수 < 90 : B등급  
70 <= 점수 < 80 : C등급  
      점수 < 70 : D등급 

```python
class_room_data = {
    'stu_no' : [1, 2, 3, 4, 5],
    'stu_name' : ['김자바', '이자바', '박자바', '최자바', '정자바'],
    'stu_score' : [85, 65, 70, 80, 95]
}

gradeList = []


for e in class_room_data['stu_score']:

    if e >= 90:
        grade = 'A'
    elif e >=80:
        grade = 'B'
    elif e >= 70:
        grade = 'C'
    else:
        grade = 'D'
        
    gradeList.append(grade)
class_room_data['stu_grade'] = gradeList

print(class_room_data)


```

7. 아래 데이터는 자바반과 자반에 소속된 학생들의 정보이다.  
나이가 30살 이상인 학생들의 점수의 합을 출력하세요.
