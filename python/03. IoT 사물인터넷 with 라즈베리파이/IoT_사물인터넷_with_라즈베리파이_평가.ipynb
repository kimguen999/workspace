### 문제 3.
df 데이터를 아래 기준으로 결측치를 처리하세요. 처리된 결과는 **df에 저장**되어야 합니다.

1. `도서분류` 컬럼 → **'미분류'** 로 채우기
2. `할인율` 컬럼 → **0** 으로 채우기
3. `판매가격` 컬럼 → **평균값** 으로 채우기
4. `수량` 컬럼 → **중앙값** 으로 채우기
5. `평점` 컬럼 → **평균값** 으로 채우기 (소수점 둘째 자리)

```python

# 1
df['도서분류'] = df['도서분류'].fillna('미분류',inplace=True)
# 2
df['할인율'] = df['할인율'].fillna(0,inplace=True)
# 3
df['판매가격'] = df['판매가격'].fillna(df['판매가격'].mean(), inplace=True)
# 4
df['수량'] = df['수량'].fillna(df['수량'].median(), inplace=True)
# 5
df['평점'] = df['평점'].fillna(round(df['평점'].mean(),2), inplace=True)

df
```
