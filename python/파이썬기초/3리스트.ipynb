```python
# 리스트(다수의 데이터를 index 번호로 관리하는 자료형)
# 변수명에 list는 사용 금지!!!!!!!!!!!!!
list1 = [] # 빈 리스트 생성
list2 = [3, 5.5, 'python'] # 서로 다른 자료형 저장 가능

print(list1)
print(list2)
print(list2[1])
```

```python
list1 = [1,2,3] 
# 위에서 list1을 한번 선언했기에, 값 변경만 한다.
# 변수명은 중복되지 않는다.

# 리스트에 데이터 추가하기

# 1. append() : 자바스크립트의 push()와 동일한 기능, 리스트의 마지막 요소로 추가된다.
list1.append(4) #[1,2,3,4]

# 2. insert() : 원하는 index(위치)에 데이터 추가
list1.insert(2,5) #insert(위치,데이터)  [1,2,5,3,4]

print(list1)
```

```python
# 리스트의 데이터 삭제

# 1. del 명령어 사용
del list1[1] # list1의 1번 요소 데이터를 삭제
print(list1)

# 주의사항!!!!!!!!!
# del list1   -------   list1이라는 변수를 통으로 날려버리므로 주의!!!!!!!!!!!

```

```python
# 리스트 데이터 수정

# list2의 1번 요소를 'hello'로 변경
list2[1] = 'hello'
print(list2)
```

```python
# 리스트 합치기
list11 = [1,2,3]
list22 = [4,5,6]

list33 = list11 + list22
# [1,2,3,4,5,6]
print(list33)

list44 = list11 * 3  # 리스트를 몇번 나열
print(list44)
```

```python
# 리스트의 슬라이싱 연산(리스트의 특정 부분만 추출)
# [시작index : 마지막 index] (마지막 index는 포함되지 않음)
lista = [1,2,3,4,5]
print(lista[1:4])

# lista 배열에서 [2,3] 데이터만 갖는 새로운 리스트를 생성
listb = lista[1 : 3]
print(listb)

listc = [1,2,3,4,5]
print(listc[2:]) # 2번 index부터 끝까지

print(listc[:3]) # 처음부터 3번 index 전까지.

print(listc[:]) # 처음부터 끝까지

print(listc[2:-1]) # 2번부터 -1번요소 전까지. 음수는 순서가 거꾸로이다.
```
